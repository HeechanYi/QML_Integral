# Benchmark results

Compare integral MAE across the ten training seeds on each fixed dataset. Run from the repository root or this notebook’s directory.

```python
from pathlib import Path
import csv
import matplotlib.pyplot as plt

root = next(p for p in (Path.cwd(), Path.cwd().parent) if (p / 'Data File' / 'Benchmark Results').is_dir())
rows = []
for filename in ['qnn_per_seed.csv', 'qsp_dqc1_seed_runs.csv']:
    with (root / 'Data File' / 'Benchmark Results' / filename).open() as stream:
        rows.extend(csv.DictReader(stream))
```

```python
model, benchmark, loss = 'QNN', 'CPF', 'MSE'
fig, ax = plt.subplots(figsize=(5, 3))
for sampling in ['Uni', 'GS', 'HMC']:
    selected = [r for r in rows if r.get('model', 'QNN') == model and r['function'] == benchmark
                and r['loss'] == loss and r['sampling'] == sampling]
    ax.scatter([int(r['seed']) for r in selected], [float(r['MAE']) for r in selected], label=sampling)
ax.set(xlabel='Training seed', ylabel='Subinterval integral MAE', yscale='log', title=f'{model} / {benchmark} / {loss}')
ax.legend()
fig.tight_layout()
```
